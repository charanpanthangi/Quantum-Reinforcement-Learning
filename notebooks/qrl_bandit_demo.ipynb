# Quantum Reinforcement Learning Bandit Demo

This notebook shows a tiny experiment where a quantum policy and a classical softmax policy learn to pull the best arm in a multi-armed bandit.

## Quick intro
- **Reinforcement learning**: an agent picks an action, gets a reward, and updates its policy.
- **Bandit problem**: there is no state, only actions (arms) with fixed reward probabilities.
- **Quantum policy**: a small parameterized quantum circuit outputs a probability distribution over actions. Training updates the gate angles to favor rewarding actions.

```python
import numpy as np
import matplotlib.pyplot as plt
%config InlineBackend.figure_formats = ['svg']

from app.envs import MultiArmedBandit
from app.trainer import train_qrl, train_classical_rl, moving_average
from app.plots import plot_episode_rewards, plot_final_performance
```

```python
# Set up a simple 2-armed bandit where arm 1 is better
np.random.seed(0)
reward_probs = [0.2, 0.8]
env = MultiArmedBandit(reward_probs)
print('Reward probabilities:', reward_probs)
```

```python
# Train both agents for a small number of episodes so the demo runs fast
weights_q, rewards_q = train_qrl(env, n_episodes=200, lr=0.2)
env.reset()
prefs_c, rewards_c = train_classical_rl(env, n_episodes=200, lr=0.2)
```

```python
# Plot learning curves as SVG
plt.figure(figsize=(6,4))
plt.plot(moving_average(rewards_q, 20), label='Quantum policy')
plt.plot(moving_average(rewards_c, 20), label='Classical softmax')
plt.xlabel('Episode')
plt.ylabel('Moving average reward')
plt.title('Learning curves (bandit)')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Compare final average reward
print('Quantum average (last 20 episodes):', np.mean(rewards_q[-20:]))
print('Classical average (last 20 episodes):', np.mean(rewards_c[-20:]))
```
